```python
import pandas as pd
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('products_asos.csv', engine='python', on_bad_lines='skip')
df.head()
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
df = df.dropna(subset = ['price'])
```

```python
df.head()
```

```python
print(f'Data Loaded: {len(df)} rows')
df.head()
```

```python
df['description'] = df['description'].astype(str)
def get_brand(text):
  if 'by' in text:
    try:
     return text.split('by ')[1].strip().split(' ')[0]
    except (IndexError, ValueError):
      return "unknown"
  return "unknown"
df['brand'] = df['description'].apply(get_brand)
```

```python
brand_map = {
    'New': 'New Look',
    'River': 'River Island',
    'Miss': 'Miss Selfridge',
    'Topshopwelcome': 'Topshop'

}
df['Brand'] = df['brand'].map(brand_map).fillna(df['brand'])

brand_counts = df['Brand'].value_counts()
valid_brands = brand_counts[brand_counts >= 5].index
df_clean = df[df['Brand'].isin(valid_brands)].copy()

print(df_clean['Brand'].value_counts().head(5))
```

```python
df.head()
```

```python
# 1.Function to analyze stockouts
def calculate_phantom_revenue(size_str):
  if not isinstance(size_str, str):
    return 0, 0.0

  # Split by comma to get individual size entries
  sizes = [s.strip() for s in size_str.split(',')]
  total_sizes = len(sizes)

  if total_sizes == 0:
    return 0, 0.0

  out_of_stock_count = 0

  # Check if ' - Out of stock' phrase indicates general stockout for listed sizes
  if ' - Out of stock' in size_str:
      if size_str.strip().endswith(' - Out of stock'):
          # If it ends with ' - Out of stock', it implies all preceding sizes are out of stock
          sizes_only_str = size_str.rsplit(' - Out of stock', 1)[0].strip()
          if sizes_only_str: # Ensure there are actual sizes listed
              out_of_stock_count = len([s.strip() for s in sizes_only_str.split(',') if s.strip()])
          elif size_str.strip() == 'Out of stock': # Handle case where string is just 'Out of stock'
              out_of_stock_count = 1
          else:
              out_of_stock_count = 0 # Fallback, unlikely for the given data
      else:
          # If ' - Out of stock' is present but not at the end, it's for individual size entries
          out_of_stock_count = sum(1 for s_entry in sizes if 'Out of stock' in s_entry)
  else:
      # No 'Out of stock' phrase found, so no stockouts
      out_of_stock_count = 0

  rate = out_of_stock_count / total_sizes if total_sizes > 0 else 0.0

  return out_of_stock_count, rate

metrics = df_clean['size'].apply(lambda x: calculate_phantom_revenue(x))

df_clean['stockout_Count'] = [x[0] for x in metrics]
df_clean['stockout_Rate'] = [x[1] for x in metrics]

df_clean['Lost_revenue'] = df_clean['price'] * df_clean['stockout_Count']

cols = ['Brand','name','price','stockout_Count','Lost_revenue']
print(df_clean[cols].sort_values(by='Lost_revenue', ascending=False).head(5)[cols])
```

```python
Brand_strategy = df_clean.groupby('Brand').agg({
    'price': 'mean',
    'stockout_Rate': 'mean',
    'Lost_revenue': 'sum',
    'name': 'count'
}).reset_index()

Brand_strategy = Brand_strategy[Brand_strategy['name'] > 10]

plt.figure(figsize=(12,8))
sns.scatterplot(
    data=Brand_strategy,
    x='price',
    y='stockout_Rate',
    size='Lost_revenue',
    sizes=(50,500),
    alpha=0.7,
    palette='viridis'
)

winners = Brand_strategy[
    (Brand_strategy['price'] > 40) &
    (Brand_strategy['stockout_Rate'] < 0.4)
]

for i in range(len(winners)):
  plt.text(
      winners.iloc[i]['price']+1,
      winners.iloc[i]['stockout_Rate'],
      winners.iloc[i]['Brand']
  )

plt.title('Brand strategy Analysis')
plt.xlabel('Average Price')
plt.ylabel('Stockout Rate')
plt.axvline(x=40, color='red', linestyle='--')
plt.axhline(y=0.4, color='red', linestyle='--')
plt.show()
```
